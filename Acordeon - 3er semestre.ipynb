# Manejo de columnas

Es posible seleccionar una o varias columnas:
* Una columna
* Dos columnas
* Una lista de columnas

La sintaxis general es `df[las columnas que quiero]`

### Ejemplo para una columna
Para seleccionar la columna `col5` usamos `df["col5"]`

```python
df["col5"]
```

### Ejemplo para dos columnas

Para seleccionar dos columnas la sintaxis general es `df[["col8","col7"]]`, la diferencia con seleccionar una columna es que en esta ocasión hay doble corchete:

```python
df[["col8","col7"]]
```

### Ejemplo para seleccioar varias columnas
En general para seleccionar varias columnas es necesario escribir la lista de las columnas y pasarla con corchetes al dataframe, por ejemplo, si queremos las columnas `col1`,`col10`,`col6` y `col8`, podemos utilizar la siguiente sintasis

```python
columnas_seleccionadas = ["col1","col10","col6","col8"]
df[columnas_seleccionadas]
```

# Filtrar datos
Al crear una condición sobre una columna estamos creando una mascara logica (una columna True o False por cada fila que cumpla la condicion), por ejemplo `df["col6"] == "Sur"` crea una columna que será True en aquellas filas donde el valor sea "Sur" y False cuando no lo sea. 

```python
df["col6"] == "Sur"
```

Para acceder a la tabla completa en la que se cumpla la condición `df["col6"] == "Sur"` sólo basta con agregarlo entre corcente en el dataframe: `df[df["col6"] == "Sur"]`. El resultado contiene las filas y columnas del dataframe en las cuales la columna `col6` tenga el valor "Sur"

```python
df[df["col6"] == "Sur"]
```

### Filtrar por más de una condición
Para filtrar por más de una condición basta con separar las condiciones por paréntesis y conectarlos con los operadores lógicos `&` (and), `|` (or), `~` (not), ejemplo:

*las filas cuya columna 6 sea "Sur" y col5 diferente de "B"
`(df["col6"]=="Sur") & (df["col5"]!="B")`

*Las filas cuya columna 5 sea "A" o columna 6 sea "Norte"
`(df["col5"]=="A") | (df["col6"] == "Norte")

```python
df[(df["col6"]=="Sur") & (df["col5"]!="B")]
```

# Ordenar Datos
Ordenaremos el DataFrame utilizando una o varias columnas. Ya sea de menor a mayor o de mayor a menor


```python
df.sort_values("col1") #Menor a mayor (esto ya es por defecto)
```

```python
df.sort_values("col1", ascending = False) #De mayor a menor
```

# Estadisticas descriptivas

Para obtener la estadistica de una columna basta con seleccionar la columna y aplicar el metodo, por ejemplo la media es `.mean()`, el maximo `.max()`, la mediana `.median()`, por ejemplo 

* `df["col1"].mean()` Calcula la media


```python
df["col1"].mean()
```

# Trabajar con variables categoricas
Para analizar variables de texto o categorías, utilizamos `value_counts()` para contar cuántas veces aparece cada valor, `unique()` para obtener una lista de los valores distintos (osea que variables hay), o `nunique()` para saber la cantidad exacta de valores diferentes (cuantas variables hay).

```python
conteo_categorías = df["col10"].value_counts()
display(conteo_categorías)
```

```python
valores_unicos = df["col9"].unique()
valores_unicos
```

```python
valores_no_unicos = df["col9"].nunique()
valores_no_unicos
```

# .groupby

```python
df.groupby("col10")[["col1","col3"]].mean()
```

```python
df.groupby(["col10","col5","col9"])[["col1","col3"]].mean()
```

# Crear y modificar columnas
Para hacer una columna nueva, solo pones el nombre que quieras entre corchetes y le dices a qué va a ser igual. Puede ser una operación con otras columnas (como sumarlas, que es lo que hacemos en el ejemplo) o meterle un valor fijo a todas las filas.

```python
df["suma_col1_col3"] = df["col1"] + df["col3"]
df[["col1", "col3", "suma_col1_col3"]].head()
```

# Combinar operaciones
Por ejemplo aqui separamos solo las filas que digan `"sur"`, de ahí obtenemos el promedio de la `"col1"` por cada categoria de la `"col10"` y la ordenamos de mayor a menor 

```python
resultado_final = df[df["col6"] == "Sur"].groupby("col10")[["col1"]].mean().sort_values("col1",ascending=False)
resultado_final
```
