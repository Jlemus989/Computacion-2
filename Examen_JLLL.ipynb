```python
import pandas as pd

orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
geo = pd.read_csv("olist_geolocation_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")

# Convertir las columnas de fechas de orders

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)
```

# NIVEL 3
#### 1. En la tabla olist_order_items_dataset, crea una columna llamada costo_total como la suma de price y freight_value. Calcula posteriormente el costo total promedio de los artículos.

```python
order_items["costo_total"] = order_items["price"] + order_items["freight_value"]
order_items["costo_total"].mean()
```

# NIVEL 3 
#### 2. Agrupa los pedidos de olist_orders_dataset por su estado y calcula cuántos pedidos existen en cada estado. Ordena el resultado de mayor a menor cantidad de pedidos

```python
df2_nivel3_p1 = orders.merge(order_items, on = "order_id", how = "left")
df2_nivel3p2 = df2_nivel3_p1.merge(customers, on = "customer_id", how = "left")
df2_nivel3p3 = df2_nivel3p2.groupby("customer_state")[["order_id"]].count().reset_index()
df2_nivel3p4 = df2_nivel3p3.sort_values("order_id", ascending = False)
df2_nivel3p4
```

# NIVEL 5
#### Determina, para cada estado de Brasil, el número de pedidos realizados y el precio promedio de los artículos correspondientes a dichos pedidos

```python
df1_nivel5 = orders.merge(order_items, on = "order_id", how = "left")
df1_nivel5_paso2 = df1_nivel5.merge(customers, on = "customer_id", how = "left")
df1_nivel5_paso2.groupby("customer_state").agg({"order_id":"count","price":"mean"})
```

# NIVEL 5
#### Identifica los 10 clientes que hayan generado el mayor gasto total en productos. Para cada cliente muestra su identificador, ciudad,  estado y gasto tota

```python
#Agarro el merge hecho en el problema anterior porque ya están juntos orders, order items y customers
df2_nivel5_paso1 = df1_nivel5_paso2.groupby("customer_unique_id")[["price"]].sum().reset_index().sort_values("price",ascending = False).head(10)
df2_nivel5_paso1.merge(customers, on = "customer_unique_id", how = "inner").sort_values("price", ascending = False)[["customer_unique_id","customer_city","customer_state","price"]]
```

# NIVEL 5
#### Determina la puntuación promedio de las reseñas para cada categoría de product

```python
df3_paso2 = order_reviews.merge(order_items, on = "order_id", how = "left")
df3_paso3 = df3_paso2.merge(products, on = "product_id", how = "left")
df3_paso3.groupby("product_category_name")[["review_score"]].mean()
```
