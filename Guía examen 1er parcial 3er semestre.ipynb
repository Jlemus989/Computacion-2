# 41. Para cada estado del cliente, calcula el número de pedidos entregados y el porcentaje que representan respecto del total de pedidos de ese estado

```python
orders.columns
```

```python
customers.columns
```

```python
paso1 = orders.merge(customers, on = "customer_id", how = "left")
paso1["entregados"] = paso1["order_status"] == "delivered"
paso2 = paso1.groupby("customer_state").agg({"entregados":["sum","count"]})
paso2["porcentaje"] = ((paso2["entregados","sum"])/(paso2["entregados","count"]))*100
paso2[[("entregados","sum"),("porcentaje","")]]
```

# 42. Encuentra las 10 categorías con mayor número de artículos vendidos. Une ítems con productos y excluye categorías nulas

# 44. Calcula el gasto total por cliente único sumando el precio y el flete de sus artículos. Une ítems con clientes a través de pedidos y ordena de mayor a menor

```python
paso1 = orders.merge(order_items, on = "order_id", how = "left")
paso2 = paso1.merge(customers, on = "customer_id", how = "left")
paso2.groupby("customer_unique_id")[["valor_total_items"]].sum().sort_values("valor_total_items", ascending = False)
```
