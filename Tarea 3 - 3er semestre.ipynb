# 15.- ¿Cuál es la película con mayor duración?

```python
df.groupby("duracion_min").max()[["titulo"]]
#La pelicula con más duración es la ultima de la lista en este caso
```

# 16.- ¿Cuál es la película con mayor rating?

```python
df[["titulo","rating"]].sort_values("rating", ascending = False)
#Esta es otra forma de resolver la anterior pero ahora con la columna "rating", pero como todas tiene 0 de rating pues no se ve
```

# 17.- ¿Cuál es la película que recibió más votos?

```python
df[["titulo","votos"]].sort_values("votos", ascending = True)
#de igual forma, todos tienen 0 votos, entonces ninguna tiene mas votos
```

# 18.- ¿Qué géneros aparecen en el dataset y cuántas películas hay de cada género?

```python
df.groupby("genero")[["titulo"]].count()
```

# 19.- ¿Qué país aparece con mayor frecuencia?

```python
df.groupby("idioma_original")["titulo"].count()
#quiero creer que se refería al idioma, si es así, el inglés es el que aparece con mayor frecuencia
```
